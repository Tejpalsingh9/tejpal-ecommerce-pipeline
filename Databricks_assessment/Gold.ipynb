```python
%sql
CREATE OR REPLACE VIEW de_workspace26.tejpal_shop.tejpal_gold_monthly_revenue AS
SELECT 
  year(order_date) as year,
  month(order_date) as month,
  region,
  SUM(revenue) as total_revenue
FROM de_workspace26.tejpal_shop.tejpal_silver_orders
GROUP BY year(order_date), month(order_date), region;
```

```python
%sql
CREATE OR REPLACE VIEW de_workspace26.tejpal_shop.tejpal_gold_top_products AS
SELECT *,
       RANK() OVER (PARTITION BY category ORDER BY total_revenue DESC) as rank
FROM (
  SELECT 
    product_name,
    category,
    SUM(revenue) as total_revenue
  FROM de_workspace26.tejpal_shop.tejpal_silver_orders
  GROUP BY product_name, category
)
```
