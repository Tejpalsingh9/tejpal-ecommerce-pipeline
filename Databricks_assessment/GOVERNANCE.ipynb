```python
%sql
CREATE OR REPLACE FUNCTION de_workspace26.tejpal_shop.mask_email(email STRING)
RETURN CASE 
  WHEN email IS NULL THEN NULL
  ELSE concat(substr(email, 1, 2), '****')
END;
```

```python
%sql
ALTER TABLE de_workspace26.tejpal_shop.tejpal_silver_customers
ALTER COLUMN email
SET MASK de_workspace26.tejpal_shop.mask_email;
```
